# ZenteiQ MaxText Runner

Use a fresh Colab runtime once per backend. Change the runtime type to match `BACKEND`, then run the cells in order.

```python
# Choose exactly one backend for this Colab runtime.
BACKEND = "gpu"  # "cpu", "gpu", or "tpu"

# Paste your public GitHub repo URL here when opening this notebook directly in Colab.
REPO_URL = "https://github.com/exkzit-code/zenteiq-assignment-15062026.git"

from pathlib import Path
import os
import subprocess

repo_dir = Path("/content/zenteiq-maxtext-assignment")
if not Path("scripts").exists():
    if not REPO_URL:
        raise ValueError("Set REPO_URL to your public GitHub repo URL, then rerun this cell.")
    if repo_dir.exists():
        subprocess.run(["git", "-C", str(repo_dir), "pull", "--ff-only"], check=True)
    else:
        subprocess.run(["git", "clone", REPO_URL, str(repo_dir)], check=True)
    os.chdir(repo_dir)
elif Path(".git").exists():
    subprocess.run(["git", "pull", "--ff-only"], check=True)

print("Backend:", BACKEND)
print("Working directory:", Path.cwd())
```

## Setup

Use a fresh runtime per backend because the MaxText extras can differ for CPU, GPU, and TPU.

```python
import subprocess

subprocess.run(["bash", "scripts/install_maxtext_colab.sh", BACKEND], check=True)
```

## Estimate Planned Parameter Sizes

```python
subprocess.run(["python3", "scripts/estimate_params.py"], check=True)
```

## Run MaxText

```python
import json
from pathlib import Path

matrix = json.loads(Path("configs/runs.json").read_text())
expected = [run["run_id"] for run in matrix["runs"] if run["backend"] == BACKEND]
for run_id in expected:
    subprocess.run(["python3", "scripts/run_maxtext_colab.py", "--run-id", run_id], check=True)

print("Completed:", expected)
```
