# ZenteiQ MaxText Runner

Use this notebook once per Colab backend. Change the Colab runtime type to match `BACKEND`, then run the cells in order.

```python
# Choose exactly one backend for this Colab runtime.
BACKEND = "gpu"  # "cpu", "gpu", or "tpu"

# Paste your public GitHub repo URL here when opening this notebook directly in Colab.
REPO_URL = "https://github.com/exkzit-code/zenteiq-assignment-15062026.git"

from pathlib import Path
import os
import subprocess

repo_dir = Path("/content/zenteiq-maxtext-assignment")
if not Path("scripts").exists():
    if not REPO_URL:
        raise ValueError("Set REPO_URL to your public GitHub repo URL, then rerun this cell.")
    if repo_dir.exists():
        subprocess.run(["git", "-C", str(repo_dir), "pull", "--ff-only"], check=True)
    else:
        subprocess.run(["git", "clone", REPO_URL, str(repo_dir)], check=True)
    os.chdir(repo_dir)
elif Path(".git").exists():
    subprocess.run(["git", "pull", "--ff-only"], check=True)

print("Backend:", BACKEND)
print("Working directory:", Path.cwd())
```
